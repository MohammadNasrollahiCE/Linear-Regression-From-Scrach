# Analysing and Cleaning the California Housing dataset

#### Importing needed packages

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Reading the csv formated dataset 

```python
from sys import path
path.append('..')

df = pd.read_csv("../data/housing.csv")
df
```

#### Aplying some statistical analysis

```python
df.describe()
```

```python
df['rooms_per_householde'] = df["total_rooms"] / df['households']
df.head()
```

```python
df['bedrooms_per_room'] = df['total_bedrooms'] / df["total_rooms"]
df.head()
```

```python
df['population_per_housing'] = df["population"] / df['households']
df.head()
```

```python
df.drop('total_rooms', axis=1, inplace= True)
df.drop('total_bedrooms', axis=1, inplace= True)
df.head()
```

#### One-Hot Encoding on "ocean_proximity"

```python
df['ocean_proximity'].unique()
```

```python
encoded = pd.get_dummies(df['ocean_proximity'], dtype= int)
df = pd.concat([df, encoded], axis=1)
df.drop('ocean_proximity', axis=1, inplace= True)
df.drop('<1H OCEAN', axis=1, inplace= True)
df.head()
```

#### Handling the missing values and duplicated rows

```python
df.info()
```

```python
df.duplicated().sum()
```

#### data has no any duplicated row

```python
df.isna().sum()
```

```python
df['bedrooms_per_room'] = df['bedrooms_per_room'].fillna(df['bedrooms_per_room'].mean())
```

```python
df.isna().sum()
```

#### missing values removed

```python
df.describe()
```

#### Given the outliers revealed by the box plot, as well as the significant skewness observed in the histogram for the 'population' column, I will apply a log transformation and add the resulting values ​​as a new column to the dataframe; this will allow me to evaluate the impact of both versions during model training.

```python
df['population_log'] = np.log1p(df['population'])
df.head()
```

#### Now, I will drow box plot and histogram again 

```python
sns.boxplot(data= df['population_log'])
```

```python
sns.histplot(data= df['population_log'])
```

```python
Q1 = df['population_log'].quantile(0.25)
Q3 = df['population_log'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print('Lower Bound:', lower_bound)
print('Upper Bound:', upper_bound)

print('Lower Outliers:',
      (df['population_log'] < lower_bound).sum())

print('Upper Outliers:',
      (df['population_log'] > upper_bound).sum())
```

```python
print(np.expm1(5.493402417179638))
print(np.expm1(8.629657544321613))
```
